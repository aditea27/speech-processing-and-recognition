```python
import numpy as np
import random

# ------------------------------------------------------------
# (a) Define HMM Parameters
# ------------------------------------------------------------

# Hidden states (phonemes)
states = ['/s/', '/p/', '/iː/', '/tʃ/']

# Observations (features)
observations = ['Energy', 'Pitch', 'Duration']

# Initial probabilities (start always at /s/)
initial_prob = {
    '/s/': 1.0,
    '/p/': 0.0,
    '/iː/': 0.0,
    '/tʃ/': 0.0
}

# Transition Matrix
transition_prob = {
    '/s/': {'/s/': 0.1, '/p/': 0.8, '/iː/': 0.1, '/tʃ/': 0.0},
    '/p/': {'/s/': 0.0, '/p/': 0.1, '/iː/': 0.8, '/tʃ/': 0.1},
    '/iː/':{'/s/': 0.0, '/p/': 0.0, '/iː/': 0.2, '/tʃ/': 0.8},
    '/tʃ/':{'/s/': 0.2, '/p/': 0.0, '/iː/': 0.0, '/tʃ/': 0.8}
}

# Emission probabilities (Energy, Pitch, Duration)
emission_prob = {
    '/s/':  {'Energy': 0.7, 'Pitch': 0.2, 'Duration': 0.1},
    '/p/':  {'Energy': 0.5, 'Pitch': 0.3, 'Duration': 0.2},
    '/iː/': {'Energy': 0.3, 'Pitch': 0.5, 'Duration': 0.2},
    '/tʃ/': {'Energy': 0.4, 'Pitch': 0.4, 'Duration': 0.2}
}

# ------------------------------------------------------------
# (b) Display function
# ------------------------------------------------------------
def display_hmm():
    print("\n=== INITIAL PROBABILITIES ===")
    for s, p in initial_prob.items():
        print(f"{s} : {p}")

    print("\n=== TRANSITION MATRIX ===")
    for s_from, row in transition_prob.items():
        print(f"{s_from} -> {row}")

    print("\n=== EMISSION MATRIX ===")
    for s, row in emission_prob.items():
        print(f"{s} -> {row}")


# ------------------------------------------------------------
# Helper function: weighted random choice
# ------------------------------------------------------------
def weighted_choice(prob_dict):
    states = list(prob_dict.keys())
    probs = list(prob_dict.values())
    return random.choices(states, weights=probs, k=1)[0]

# ------------------------------------------------------------
# (c) Generate one phoneme and observation sequence
# ------------------------------------------------------------
def generate_sequence(length=4):
    sequence = []
    observation_sequence = []

    # Start at /s/
    current_state = '/s/'
    sequence.append(current_state)

    # Generate sequence
    for _ in range(length - 1):
        current_state = weighted_choice(transition_prob[current_state])
        sequence.append(current_state)

    # Generate observations for each phoneme
    for state in sequence:
        chosen_obs = weighted_choice(emission_prob[state])
        observation_sequence.append(chosen_obs)

    return sequence, observation_sequence


# ------------------------------------------------------------
# MAIN EXECUTION
# ------------------------------------------------------------
display_hmm()
phonemes, observations_out = generate_sequence()

print("\nGenerated Phoneme Sequence:")
print(phonemes)

print("\nGenerated Acoustic Observations:")
print(observations_out)

# ------------------------------------------------------------
# (d) Inference
# ------------------------------------------------------------
print("\n--- INFERENCE ---")
print("""
This HMM simulates pronunciation of the word 'speech'.
Starting from /s/, transitions follow the highest probabilities:
    /s/ → /p/ → /iː/ → /tʃ/
The emission probabilities generate likely acoustic features
for each phoneme such as Energy, Pitch, or Duration.

The generated sequence mimics how phonemes and acoustic features
might unfold in real speech processing.
""")
```
